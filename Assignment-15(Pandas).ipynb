```python
import pandas as pd

data = {
    'ProductID': [101, 102, 103, 104, 105, 106],
    'ProductName': ['Laptop', 'Mobile', 'Headphones', 'Keyboard', 'Monitor', 'Mouse'],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Accessories', 'Electronics', 'Accessories'],
    'UnitsSold': [100, 200, 180, 120, 80, 250],
    'UnitPrice': [60000, 30000, 2000, 1500, 12000, 500]
}

df = pd.DataFrame(data)

# Calculate revenue
df['Revenue'] = df['UnitsSold'] * df['UnitPrice']

# 1. Highest revenue product
max_index = df['Revenue'].idxmax()

print("Highest Revenue Product:", df.loc[max_index, 'ProductName'])
print("Total Revenue:", df.loc[max_index, 'Revenue'])

# 2. Average unit price for Electronics
avg_price = df.loc[
    df['Category'] == 'Electronics',
    'UnitPrice'
].mean()

print("Average Electronics Unit Price:", avg_price)

# 3. Products with more than 150 units sold
products = df.loc[df['UnitsSold'] > 150, 'ProductName']

print("Products with more than 150 units sold:")
print(products.to_list())
```

```python
# 29.	Create a DataFrame with the following columns: StudentID, StudentName, Math, Science, and English. Include at least 5 students
#○	Identify the student with the highest average score across all subjects and provide their name and average score.
#○	Calculate the average Math score for all students and list students who scored above this average.
#○	Find the subject in which students scored the highest average overall.

data = {
    'StudentID': [101, 102, 103, 104, 105],
    'StudentName': ['Arun', 'Priya', 'Rahul', 'Sneha', 'Kiran'],
    'Math': [85, 90, 78, 95, 88],
    'Science': [90, 85, 92, 88, 80],
    'English': [80, 95, 85, 90, 92]
}

df = pd.DataFrame(data)

print(df)
```

```python
math_avg = df['Math'].mean()
print("Average Math Score:", math_avg)
above_avg = df[df['Math'] > math_avg]
print(above_avg[['StudentName', 'Math']])
```

```python
subject_avg = df[['Math', 'Science', 'English']].mean()
print(subject_avg)
```

```python
highest_subject = subject_avg.idxmax()
print("Highest Average Subject:", highest_subject)
print("Average Score:\n", subject_avg)
```
